```sql
SELECT *,
       ROW_NUMBER() OVER (
           PARTITION BY department
           ORDER BY sales_amount DESC
       ) AS rn
FROM employee_sales;
```

```sql
SELECT *,
       RANK() OVER (
           PARTITION BY department
           ORDER BY sales_amount DESC
       ) AS rnk
FROM employee_sales;
```

```sql
SELECT *,
       DENSE_RANK() OVER (
           PARTITION BY department
           ORDER BY sales_amount DESC
       ) AS drnk
FROM employee_sales;
```

```sql
SELECT *,
       NTILE(3) OVER (
           PARTITION BY department
           ORDER BY sales_amount DESC
       ) AS bucket
FROM employee_sales;
```

```sql
SELECT *,
       SUM(sales_amount) OVER (
           PARTITION BY department
       ) AS dept_total
FROM employee_sales;
```

```sql
SELECT *,
       AVG(sales_amount) OVER (
           PARTITION BY department
       ) AS dept_avg
FROM employee_sales;
```

```sql
SELECT *,
       COUNT(*) OVER (
           PARTITION BY department
       ) AS dept_count
FROM employee_sales;
```

```sql
SELECT *,
       MIN(sales_amount) OVER (
           PARTITION BY department
       ) AS dept_min
FROM employee_sales;
```

```sql
SELECT *,
       MAX(sales_amount) OVER (
           PARTITION BY department
       ) AS dept_max
FROM employee_sales;
```

```sql
SELECT *,
       LAG(sales_amount,1) OVER (
           PARTITION BY department
           ORDER BY sales_amount DESC
       ) AS previous_sales
FROM employee_sales;
```

```sql
SELECT *,
       LEAD(sales_amount,1) OVER (
           PARTITION BY department
           ORDER BY sales_amount DESC
       ) AS next_sales
FROM employee_sales;
```

```sql
SELECT *,
       FIRST_VALUE(sales_amount) OVER (
           PARTITION BY department
           ORDER BY sales_amount DESC
       ) AS highest_sales
FROM employee_sales;
```

```sql
SELECT *,
       PERCENT_RANK() OVER (
           PARTITION BY department
           ORDER BY sales_amount DESC
       ) AS percent_rank
FROM employee_sales;
```
