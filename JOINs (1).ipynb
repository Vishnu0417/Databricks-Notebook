```sql
CREATE OR REPLACE TABLE table_a (
  id INT,
  name STRING,
  age INT,
  city STRING
);

INSERT INTO table_a VALUES
  (1, 'Alice', 30, 'New York'),
  (2, 'Bob', 25, 'Los Angeles'),
  (3, 'Charlie', 35, 'Chicago'),
  (4, 'Diana', 28, 'Houston');

CREATE OR REPLACE TABLE table_b (
  id INT,
  department STRING,
  salary INT,
  country STRING
);

INSERT INTO table_b VALUES
  (2, 'HR', 60000, 'USA'),
  (3, 'Engineering', 90000, 'USA'),
  (5, 'Marketing', 50000, 'Canada'),
  (6, 'Sales', 55000, 'UK');
```

```sql
select* from table_b;
```

```sql
select * from table_a;
```

```sql
SELECT a.id, a.name, b.department, b.salary
FROM table_a a
INNER JOIN table_b b
ON a.id = b.id;
```

```sql
SELECT a.*, b.department, b.salary
FROM table_a a
LEFT JOIN table_b b
ON a.id = b.id;
```

```sql
SELECT a.name, b.*
FROM table_a a
RIGHT JOIN table_b b
ON a.id = b.id;
```

```sql
SELECT *
FROM table_a a
FULL OUTER JOIN table_b b
ON a.id = b.id;
```

```sql
SELECT *
FROM table_a
CROSS JOIN table_b;
```

```sql
SELECT a1.id AS id1, a1.name AS name1, a2.id AS id2, a2.name AS name2
FROM table_a a1
INNER JOIN table_a a2
ON a1.city = a2.city AND a1.id <> a2.id;
```

```sql
SELECT a1.id AS id1, a1.name AS name1, a2.id AS id2, a2.name AS name2
FROM table_a a1
INNER JOIN table_a a2
ON a1.age = a2.age AND a1.id <> a2.id;
```
